```python
def reg_per_season(season):
    season_df = master95_15[
        (master95_15['SalYear'] == season) & (master95_15['Free'] == 1)
    ]

    model = smf.ols(formula='lnSal ~ BA + ISO + Eye + PA + Exp + Exp2 + C(POS)', data=season_df).fit()

    return summary_col([model], stars=True, info_dict={"no_obvs": lambda x: f"{int(x.nobs)}"}).tables[0]
```

```python
season_list = sorted(master95_15['SalYear'].unique())

tables_dict = {}

for season in season_list:
    df = reg_per_season(season=season)
    tables_dict[season] = df.rename(columns={'lnSal': str(season)})

regressions_by_season = pd.concat(tables_dict.values(), axis=1)

regressions_by_season = regressions_by_season.reset_index()
regressions_by_season = regressions_by_season.rename(columns={'index': 'variable'})
```

```python
pre_mb = regressions_by_season.iloc[:, 0:7]
moneyball_era = regressions_by_season.iloc[:, [0] + list(range(7, 15))]
post_mb= regressions_by_season.iloc[:, np.r_[0, 15 : len(regressions_by_season.columns)-1]]
```

Determine the season with the largest “ISO” coefficient in each era.

```python
eras = [pre_mb, moneyball_era, post_mb]

for era in eras:
    values = era.loc[
        era['variable'] == 'ISO',
    ].values[0][1:]

    clean_values = [float(v.replace('*', '')) for v in values]
    years = list(era.columns)
    year_vals_dict = dict(zip(years[1:], clean_values))

    max_iso = max(clean_values)
    year = ''

    for y, val in year_vals_dict.items():
        if val == max_iso:
            year = y

    print(f'{year}:  {max_iso}')
```

How many times is “Eye” significant at the .05 level in each era respectively?

```python
for era in eras:
    row = era.loc[era['variable'] == 'Eye'].values[0]

    significant__05 = 0

    for val in row[1:]:
        asterics_count = val.count('*')
        if asterics_count >= 2:
            significant__05 += 1

    print(f'Significant {significant__05} times')
```

# Running the Pooled Regression  
Subset data to only include free agent players and add a “PreMB” dummy variable to your dataframe for seasons prior to 2004.  

Run a pooled regression for free agent players using the format same model structure developed in your function in 6).  Your pooled regression should have the form: lnSal ~ (model from 6)) + Pre-MB*(model from 6))

```python
free_agents = master95_15[master95_15['Free'] == 1].copy()
free_agents['PreMB'] = (free_agents['SalYear'] < 2004).astype(int)

pooled_formula = (
    'lnSal ~ BA + ISO + Eye + PA + Exp + Exp2 + C(POS) '
    '+ PreMB*(BA + ISO + Eye + PA + Exp + Exp2 + C(POS))'
)

pooled_regression = smf.ols(formula=pooled_formula, data=free_agents).fit()
pooled_regression.summary()
```
